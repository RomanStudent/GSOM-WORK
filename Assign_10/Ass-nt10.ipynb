```python
import shutil
import os 
with open("source.txt", "w") as f:
    f.write("This is a test file.")
os.makedirs("destination_folder", exist_ok=True)

shutil.copy("source.txt", "destination_folder/copied.txt")
print("File copied using shutil.copy()")

shutil.move("destination_folder/copied.txt", "moved_file.txt")
print("File moved using shutil.move()")

usage = shutil.disk_usage(".")
print(f"Disk usage (Total, Used, Free): {usage} bytes")

shutil.make_archive("my_archive", "zip", "destination_folder")
print("Archive created using shutil.make_archive()")

shutil.copytree("destination_folder", "copied_tree")
print("Directory copied using shutil.copytree()")
```

```python
os.remove("source.txt")
os.remove("moved_file.txt")
os.remove("my_archive.zip")
shutil.rmtree("destination_folder") 
shutil.rmtree("copied_tree")
```

```python
import time

seconds = time.time()
print(f"Seconds since epoch: {seconds}")

print(f"Local time (string): {time.ctime()}")

local_struct = time.localtime(seconds)
print(f"Local time (struct_time): {local_struct}")

utc_struct = time.gmtime(seconds)
print(f"UTC time (struct_time): {utc_struct}")

print(f"Converted back to string: {time.asctime(local_struct)}")

back_to_seconds = time.mktime(local_struct)
print(f"Converted back to seconds: {back_to_seconds}")
```
